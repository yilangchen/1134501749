```python
import sqlite3
import pandas as pd
```

```python
# 连接到你的数据库
conn = sqlite3.connect('production.db')

# 查看所有表名
cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()
print("数据库中的表：")
for table in tables:
    print(f"  - {table[0]}")

conn.close()
```
